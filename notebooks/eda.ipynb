## Target distribution

```python
target_counts = train["Will_Buy_EV"].value_counts().reset_index()
target_counts.columns = ["Will_Buy_EV", "Count"]

fig = px.bar(
    target_counts,
    x="Will_Buy_EV",
    y="Count",
    color="Will_Buy_EV",
    title="Distribution of EV Purchase Outcome",
    text="Count",
)

fig.update_layout(
    xaxis_title="Will Buy EV", yaxis_title="Number of Customers", showlegend=False
)

fig.write_image(FIGURES_PATH / "target_distribution.png")
fig.show()
```

## Annual income distribution

```python
fig = px.histogram(
    train,
    x="Annual_Income_USD",
    color="Will_Buy_EV",
    marginal="box",
    nbins=60,
    barmode="overlay",
    opacity=0.7,
    title="Annual Income Distribution by EV Purchase Outcome",
)

fig.update_layout(xaxis_title="Annual Income (USD)", yaxis_title="Number of Customers")

fig.write_image(FIGURES_PATH / "annual_income_distribution.png")
fig.show()
```

## Age distribution by target

```python
fig = px.histogram(
    train,
    x="Age",
    color="Will_Buy_EV",
    marginal="box",
    nbins=30,
    barmode="overlay",
    opacity=0.7,
    title="Age Distribution by EV Purchase Outcome",
)

fig.update_layout(xaxis_title="Age", yaxis_title="Number of Customers")

fig.write_image(FIGURES_PATH / "age_distribution.png")
fig.show()
```

## EV purchase rate by income

```python
income_bins = pd.cut(train["Annual_Income_USD"], bins=20)

income_purchase_rate = (
    train.assign(Income_Bin=income_bins)
    .groupby("Income_Bin", observed=True)["Will_Buy_EV"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="Purchase_Rate")
)

income_purchase_rate["Income_Bin"] = income_purchase_rate["Income_Bin"].astype(str)

fig = px.bar(
    income_purchase_rate,
    x="Income_Bin",
    y="Purchase_Rate",
    title="EV Purchase Rate by Annual Income",
    text_auto=".1f",
)

fig.update_layout(
    xaxis_title="Annual Income Range",
    yaxis_title="Purchase Rate (%)",
    xaxis_tickangle=-45,
)

fig.write_image(FIGURES_PATH / "income_distribution.png")
fig.show()
```

## Purchase rate for categorical features

```python
categorical_cols = train.select_dtypes(include=["object", "str"]).columns.tolist()
categorical_cols.remove("Will_Buy_EV")

feature = categorical_cols[0]

purchase_rate = (
    train.groupby(feature)["Will_Buy_EV"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="Purchase_Rate")
    .sort_values("Purchase_Rate", ascending=False)
)

fig = px.bar(
    purchase_rate,
    x=feature,
    y="Purchase_Rate",
    title=f"EV Purchase Rate by {feature}",
    text_auto=".1f",
)

fig.update_layout(xaxis_title=feature, yaxis_title="Purchase Rate (%)")

fig.write_image(FIGURES_PATH / "purchase_rate.png")
fig.show()
```

## Numeric feature correlation heatmap

```python
numeric_cols = train.select_dtypes(include=np.number).columns

corr = train[numeric_cols].corr()

fig = px.imshow(
    corr,
    text_auto=".2f",
    aspect="auto",
    title="Correlation Between Numeric Features",
    color_continuous_scale="RdBu_r",
    zmin=-1,
    zmax=1,
)

fig.update_layout(height=700)

fig.write_image(FIGURES_PATH / "numeric_correlation.png")
fig.show()
```
